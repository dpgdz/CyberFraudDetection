```python
from datasets import load_dataset

domains = [
    "fake_news",
    "job_scams",
    "phishing",
    "political_statements",
    "product_reviews",
    "sms",
    "twitter_rumours"
]

base_url = "https://huggingface.co/datasets/difraud/difraud/resolve/main"

data_files = {}

for d in domains:
    data_files[d] = {
        "train": f"{base_url}/{d}/train.jsonl",
        "validation": f"{base_url}/{d}/validation.jsonl",
        "test": f"{base_url}/{d}/test.jsonl",
    }

datasets = {}

for d in domains:
    datasets[d] = load_dataset("json", data_files=data_files[d])
    print(f"Loaded domain: {d}, sizes =",
          {split: len(datasets[d][split]) for split in datasets[d]})
```

```python
datasets
```

```python
from datasets import Dataset, concatenate_datasets

merged = {
    "train": [],
    "validation": [],
    "test": []
}

for domain, ds_dict in datasets.items():
    for split in ["train", "validation", "test"]:
        ds_tagged = ds_dict[split].add_column("domain", [domain] * len(ds_dict[split]))
        merged[split].append(ds_tagged)

merged_train = concatenate_datasets(merged["train"])
merged_val = concatenate_datasets(merged["validation"])
merged_test = concatenate_datasets(merged["test"])

print("Merged train:", len(merged_train))
print("Merged validation:", len(merged_val))
print("Merged test:", len(merged_test))
```

```python
merged_train.to_csv('data/train.csv', index=False)
merged_val.to_csv('data/validation.csv', index=False)
merged_test.to_csv('data/test.csv', index=False)
```
